**Что видно:**
- Пропуски есть только в `Initial hardness`.
- Доля пропусков 64,730% - это либо артефакт сбора данных, либо реальные «неизвестные» значения.
- Остальные колонки заполнены полностью.

**Что видно:**

```python
#проведем логический анализ данных
checks = {'Tempering time <= 0':            (df['Tempering time (s)'] <= 0).sum(),
    'Tempering temperature < 0':      (df['Tempering temperature (ºC)'] < 0).sum(),
    'Tempering temperature > 1000':   (df['Tempering temperature (ºC)'] > 1000).sum(),
    'Final hardness вне [0, 70]':     ((df['Final hardness (HRC) - post tempering'] < 0) |
                                       (df['Final hardness (HRC) - post tempering'] > 70)).sum(),
    'Отрицательный химсостав':        (df.filter(like='%wt').lt(0).any(axis=1)).sum(),
    'Сумма элементов > 100%':         (df.filter(like='%wt').sum(axis=1) > 100).sum()}
pd.Series(checks, name='Нарушений').to_frame()
```

**Что видно:**
- Время отпуска всегда > 0.
- Температура отпуска в разумном диапазоне.
- Твёрдость в шкале HRC (0–70) в норме.
- Химический состав не содержит отрицательных значений.
- Сумма элементов не превышает 100% ни в одной строке.

Все показатели не противоречат действительности => нет логических нарушений, что можно отнести к пропускам

## 4. Выбросы и аномалии

Анализ выбросов производим двумя способами: IQR для распределений и z-score для симметричных числовых признаков.

```python
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()

def iqr_out(series):
    q1, q3 = series.quantile([0.25, 0.75])
    iqr= q3 - q1
    lower= q1 - 1.5 * iqr
    upper= q3 + 1.5 * iqr
    return ((series < lower) | (series > upper)).sum()

out_sum= pd.DataFrame({
    'выбросов (IQR)': {col: iqr_out(df[col].dropna()) for col in num_cols},
    'доля, %': {col: round(iqr_out(df[col].dropna()) / df[col].notna().sum() * 100, 2)
    for col in num_cols}}).sort_values('доля, %', ascending=False)
out_sum
```

```python
main_cols = ['Tempering time (s)',
    'Tempering temperature (ºC)',
    'Initial hardness (HRC) - post quenching',
    'Final hardness (HRC) - post tempering']

miss_values= ['?', '', 'NA', 'N/A', 'nan', 'None', '-']

def clean_num(series):
    return pd.to_numeric(series.replace(miss_values, pd.NA))

fig, axes = plt.subplots(1, len(main_cols), figsize=(15, 4))
for ax, col in zip(axes, main_cols):
    clean_num(df[col]).dropna().plot(kind='box', ax=ax)
    ax.set_title(col, fontsize=9)
plt.suptitle('Распределение ключевых признаков', y=1.02)
plt.tight_layout()
plt.show()
```

**Что видно:**
В `Tempering time` (IQR) 306 выбросов. Значения сильно растянуты из-за разных режимов эксперимента.
Для дальнейшего моделирования такие выбросы стоит сохранять, так как они физически осмысленны.

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 8))
for ax, col in zip(axes.flat, main_cols):
    df[col].dropna().hist(bins=30, ax=ax, edgecolor='black', alpha=0.7)
    ax.set_title(col)
    ax.set_xlabel('значение')
    ax.set_ylabel('частота')
plt.tight_layout()
plt.show()
```

**Наблюдения:**
- `Tempering temperature` — распределение неравномерное, часть значений повторяется.
- `Final hardness` — распределение скошено в сторону средних значений, что логично для твёрдости отпущенной стали.

## 5. Метрики и корреляции

Производим анализ описательных метрик и корреляции числовых признаков.

```python
metrics = pd.DataFrame({'среднее':      df[num_cols].mean(),
    'медиана':      df[num_cols].median(),
    'std':          df[num_cols].std(),
    'min':          df[num_cols].min(),
    'max':          df[num_cols].max(),
    'CV (std/mean)': df[num_cols].std() / df[num_cols].mean()}).round(3)
metrics
```

```python
corr = df[num_cols].corr()

plt.figure(figsize=(14, 10))
sns.heatmap(corr, annot=False, cmap='RdBu_r', center=0, vmin=-1, vmax=1)
plt.title('Корреляционная матрица числовых признаков')
plt.tight_layout()
plt.show()
```

```python
target= 'Final hardness (HRC) - post tempering'
if target in corr.columns:
    top_corr = corr[target].drop(target).sort_values(key=abs, ascending=False)
    print(f'Топ корреляций с «{target}»:')
    print(top_corr.head(10).round(3))
```

**результат анализа:**
- Наибольшая по модулю корреляция с `Final hardness` — у `Tempering temperature` r ≈ −0.869. Это ожидаемо, чем выше температура отпуска, тем ниже твёрдость.
- `Tempering time` слабо коррелирует с итоговой твёрдостью r ≈ -0.180.
- Мультиколлинеарность между химическими элементами ожидаемо низкая, кроме пар C/Mn и Cr/Ni.

## Домашнее задание 6. Расширенная визуализация

Для расширенного анализа используем **Plotly Express**. Чтобы графики рендерились на GitHub, принудительно переключаем вывод в **PNG**.

```python
import plotly.express as px
import plotly.io as pio

pio.renderers.default = 'png'
pio.templates.default = 'plotly_white'
```

### Влияние температуры отпуска на твёрдость

Основная физическая зависимость в данных: с ростом температуры отпуска твёрдость падает. Построим график зависимости.

```python
fig = px.scatter(clean_df.dropna(subset=['Tempering temperature (ºC)',
                            'Final hardness (HRC) - post tempering']),
    x='Tempering temperature (ºC)',
    y='Final hardness (HRC) - post tempering',
    color='Steel type',
    hover_data=['Tempering time (s)'],
    title='Температура отпуска / итоговая твёрдость (по маркам стали)',
    labels={
        'Tempering temperature (ºC)': 'Температура отпуска, °C',
        'Final hardness (HRC) - post tempering': 'Итоговая твёрдость, HRC'},
    opacity=0.7)
fig.update_layout(height=500)
fig.show()
```

**Наблюдение:**
- Видна отрицательная зависимость: с ростом температуры отпуска твёрдость монотонно снижается.
- Заметна нелинейность падения твердости: в диапазоне 200-400 °C твёрдость падает медленно, после 400 °C — резко.

### Распределение итоговой твёрдости по маркам стали

Сравниваем распределения твёрдости между марками.

```python
fig = px.box(clean_df.dropna(subset=['Final hardness (HRC) - post tempering']),
    x='Steel type',
    y='Final hardness (HRC) - post tempering',
    color='Steel type',
    points='outliers',           # показываем только выбросы, а не все точки
    title='Распределение итоговой твёрдости по маркам стали',
    labels={
        'Steel type': 'Марка стали',
        'Final hardness (HRC) - post tempering': 'Итоговая твёрдость, HRC'})
fig.update_layout(height=500, showlegend=False)
fig.show()
```

**Наблюдение:**
- Видно различие медиан и размаха между марками: легированные стали показывают более высокую твёрдость.
- Точки за пределами усов — выбросы, но они не ошибка, это крайние режимы отпуска.

### Распределение времени отпуска

Показано, какие значения времени отпуска встречаются в данных и как они распределены по маркам стали.

```python
fig = px.histogram(clean_df.dropna(subset=['Tempering time (s)']),
    x='Tempering time (s)',
    color='Steel type',
    nbins=30,
    title='Распределение времени отпуска по маркам стали',
    labels={'Tempering time (s)': 'Время отпуска, с',
            'count': 'Число записей'})
fig.update_layout(height=500, bargap=0.05)
fig.show()
```

**Наблюдение:**
- Распределение мультимодальное: видны пики на значениях 600с, 1800с, 3600с — это стандартные режимы эксперимента.
- Наибольшее количество экспериментов поставлены на малых временных промежутках.

### Матрица парных зависимостей ключевых признаков

```python
key_features = ['Tempering temperature (ºC)',
    'Tempering time (s)',
    'Final hardness (HRC) - post tempering']

fig = px.scatter_matrix(clean_df.dropna(subset=key_features).sample(min(200,
    len(clean_df))),
    dimensions=key_features,
    color='Steel type',
    title='Матрица парных зависимостей (выборка случайные 200 строк)',
    opacity=0.6,
    height=600)
fig.update_traces(diagonal_visible=False, showupperhalf=False)
fig.show()
```

**Наблюдение:**
- Кросс-панели «температура × твёрдость» подтверждают сильную отрицательную корреляцию.
- «Время × твёрдость» — отсутствие выраженного тренда. Время при фиксированной температуре влияет слабо.

### Тепловая карта корреляций числовых признаков

```python
numeric_corr = ['Tempering temperature (ºC)',
    'Tempering time (s)','Initial hardness (HRC) - post quenching',
    'Final hardness (HRC) - post tempering',
    'C (%wt)', 'Mn (%wt)', 'Si (%wt)', 'Cr (%wt)', 'Ni (%wt)', 'Mo (%wt)']

corr = clean_df[numeric_corr].corr()

fig = px.imshow(
    corr,
    text_auto='.2f',
    color_continuous_scale='RdBu_r',
    zmin=-1, zmax=1,
    title='Тепловая карта корреляции ключевых признаков',
    aspect='auto')
fig.update_layout(height=600, width=800)
fig.show()
```

**Наблюдение:**
- Самая сильная корреляция — `Tempering temperature` × `Final hardness` = -0.87.
- `Tempering time` почти не коррелирует ни с твёрдостью, ни с температурой.
